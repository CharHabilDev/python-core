## Exercice 1

Obtiens la date et l'heure actuelles et affiche uniquement l'année.

---

````python
from datetime import datetime

current_date = datetime.now()

print(f"Année: {current_date.year}")
````

## Exercice 2

Affiche :

- le mois ;
- le jour.

---

````python
from datetime import datetime

current_date = datetime.now()

print(f"Mois: {current_date.month}\nJour: {current_date.day}")
````

## Exercice 3

Affiche :

- l'heure ;
- les minutes ;
- les secondes.

---

````python
from datetime import datetime

current_date = datetime.now()

print(f"Heure: {current_date.hour}\nMinutes: {current_date.minute}\nSecondes: {current_date.second}")
````
