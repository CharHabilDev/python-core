## Exercice 8

Réponds avec tes propres mots :

1. Quel est le rôle du module `datetime` ?
    - manipuler, représenter et comparer des dates et des heures
2. Pourquoi les dates et heures sont-elles importantes dans les applications ?
    - analyse de donnés 
    - gestion de processus et automatisation
    - reservation
    - abonnement
3. Quelle différence existe entre `date` et `datetime` ?
    - `date` génère uniquement la date alors que `datetime` ajoute l'heure
4. Cite deux exemples de programmes qui utilisent des dates.
    - une application de calendrier
    - un système de réservation
5. Pourquoi Python fournit-il un module dédié aux dates et heures ?
    - parce que les dates et les heures sont complexes à manipuler correctement. Le module datetime fournit des outils fiables pour les créer, les comparer et effectuer des calculs temporels.
