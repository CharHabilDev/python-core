## Exercice 2

Create:

- a `date`
- a `time`
- a `datetime`

Display them.

````python
from datetime import datetime, date, time

date_ = date(2025, 8, 24)
time_ = time(18, 45, 12)
datetime_ = datetime(2025, 8, 24, 18, 45, 12)

print(f"Datetime: {datetime_}\nDate: {date_}\nTime: {time_}")
````
