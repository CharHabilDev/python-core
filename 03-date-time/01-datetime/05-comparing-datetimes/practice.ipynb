## Exercice 1

Create two different `datetime` objects.

Check whether they are equal.

````python
from datetime import datetime

first_date = datetime(2025, 12, 8)
second_date = datetime(2026, 4,1)

print(f"{first_date} = {second_date} ? {first_date == second_date}")
````

## Exercice 3

Create a past date and compare it with the current date.

Determine whether it is in the past.

````python
from datetime import datetime

past_date = datetime(2025, 8, 24)
current_date = datetime.now()

print(f"{past_date} < {current_date} ? {past_date < current_date}")
````
