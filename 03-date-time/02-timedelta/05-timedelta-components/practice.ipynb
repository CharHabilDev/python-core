## Exercices

### Exercice 1

Crée :

```text
10 jours
```

avec `timedelta`.

Affiche :

```python
.days
```

````python
from datetime import timedelta

duration = timedelta(days=10)

print(duration.days)
````


---

### Exercice 2

Crée :

```text
3 heures
```

avec `timedelta`.

Affiche :

```python
.seconds
```

````python
from datetime import timedelta

hours = timedelta(hours=3)

print(f"{hours} heures = {hours.seconds} secondes")
````


---

### Exercice 3

Crée :

```text
2 jours
+
12 heures
```

Affiche :

```python
.total_seconds()
```

````python
from datetime import timedelta

duration = timedelta(days=2, hours=12)

print(f"{duration} hours = {duration.total_seconds()} seconds")
````
