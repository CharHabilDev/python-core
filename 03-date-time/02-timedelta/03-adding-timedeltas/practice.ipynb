### Exercice 1

Récupère la date actuelle.

Ajoute :

```text
7 jours
```

Affiche le résultat.

````python
from datetime import datetime, timedelta

current_date = datetime.now()
future_date = current_date + timedelta(days=7)

print(f"Current date : {current_date}")
print(f"Date in 7 days : {future_date}")
````
