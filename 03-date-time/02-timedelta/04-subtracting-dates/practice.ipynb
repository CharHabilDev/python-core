### Exercice 1

Crée deux objets `datetime` :

```text
2026-01-01
2026-01-08
```

Soustrais-les et affiche le résultat.

````python
from datetime import datetime

past = datetime(2026, 1, 1)
future = datetime(2026, 1, 8)

duration = future - past

print(f"Duration : {duration}")
````

### Exercice 3

Crée :

```text
date actuelle
```

et

```text
1er janvier 2026
```

Soustrais-les et affiche le résultat.


````python
from datetime import datetime

past = datetime(2026, 1, 1)
current_date = datetime.now()

duration = current_date - past

print(f"Duration : {duration}")
````
