### Exercice 1

Crée une date à partir du texte :

```text
"01/01/2026"
```

en utilisant `strptime()`.


````python
from datetime import datetime

str_date = "01/01/2026"

print(datetime.strptime(str_date, '%d/%m/%Y'))
````
