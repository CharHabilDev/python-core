# Exercices

### Exercice 1

Affiche une date au format :

```text
JJ/MM/AAAA
```

à partir d'un objet `datetime`.


````python
from datetime import datetime

current_date = datetime.now()

print(f"Current date : {current_date.strftime('%d/%m/%Y')}")
````


---

### Exercice 2

Affiche une date au format :

```text
AAAA-MM-JJ
```


````python
from datetime import datetime, timedelta

futur_date = datetime.now() + timedelta(weeks=4)

print(f"Futur date : {futur_date.strftime('%Y-%m-%d')}")
````

### Exercice 3

Crée un objet `datetime`, puis affiche-le avec `strftime()`.


````python
from datetime import datetime

date = datetime(2025, 4, 17)

print(f"Date : {date.strftime("%d-%m-%Y")}")
````
