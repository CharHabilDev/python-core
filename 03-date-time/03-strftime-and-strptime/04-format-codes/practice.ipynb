### Exercice 1

Affiche une date au format :

```text
jour/mois/année
```

````python
from datetime import datetime

date = datetime(2026, 5, 14)

print(date.strftime('%d/%m/%Y'))
````

### Exercice 2

Affiche uniquement l'heure au format :

```text
heure:minute:seconde
```


````python
from datetime import datetime

current_date = datetime.now()

print(current_date.strftime('%H:%M:%S'))
````

### Exercice 3

Affiche une date et une heure au format :

```text
JJ/MM/AAAA HH:MM:SS
```

````python
from datetime import datetime

current_date = datetime.now()

print(current_date.strftime('%d/%m/%Y %H:%M:%S'))
````
