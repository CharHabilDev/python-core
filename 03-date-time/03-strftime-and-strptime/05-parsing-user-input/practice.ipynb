### Exercice 1

Demande une date à l'utilisateur :

```text
JJ/MM/AAAA
```

puis convertis-la avec `strptime()`.


````python
from datetime import datetime

user_input = input("Date : (JJ/MM/AAAA) : ").strip()

date = datetime.strptime(user_input, '%d/%m/%Y')

print(date)
````
