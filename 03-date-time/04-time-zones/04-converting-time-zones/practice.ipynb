### Exercice 3

Convert the Brussels datetime to:

```text
Asia/Tokyo
```

Display both values.


````python
from datetime import datetime
from zoneinfo import ZoneInfo

brussels_time = datetime.now(ZoneInfo("Europe/Brussels"))
utc = brussels_time.astimezone(ZoneInfo("UTC"))
tokyo_time = brussels_time.astimezone(ZoneInfo("Asia/Tokyo"))

print("Time at the same moment\n")

print(f"Brussels time : {brussels_time}")
print(f"UTC time : {utc}")
print(f"Tokyo time : {tokyo_time}")
````
