### Exercice 1

Crée un datetime actuel avec :

```text
Europe/Brussels
```

Puis affiche-le.


````python
from datetime import datetime
from zoneinfo import ZoneInfo

current_date = datetime.now(ZoneInfo('Europe/Brussels'))

print(f"Current date at Brussels : {current_date}")
````
