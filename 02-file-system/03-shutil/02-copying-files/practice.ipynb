## Exercice 3

Copie :

```text
source.txt
```

vers :

```text
copy.txt
```

avec `shutil.copy()`.

`````python
from pathlib import Path
import shutil

source = Path("sandbox/source.txt")

destination = Path("sandbox/copy.txt")

shutil.copy(source, destination)
`````

## Exercice 5

Supprime :

```text
copy.txt
```

puis recrée-le avec :

```python
shutil.copy2()
```

`````python
from pathlib import Path
import shutil

path_source = Path("sandbox/source.txt")
path_copy = Path("sandbox/copy.txt")


path_copy.unlink(missing_ok=True)

# recréation
shutil.copy2(path_source, path_copy)
`````

## Exercice 6

Copie :

```text
source.txt
```

dans un dossier nommé :

```text
backup/
```

`````python
from pathlib import Path
import shutil

root = Path('sandbox')
path_source = root / "source.txt"
path_copy = root / "backup/source.txt"

path_copy.parent.mkdir(
    parents=True,
    exist_ok=True
)

shutil.copy(path_source, path_copy)
`````
