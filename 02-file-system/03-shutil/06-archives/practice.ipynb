## Exercice 1

Crée la structure suivante :
```txt
project/
├── README.md
├── main.py
└── data/
    └── users.csv
```

````python
from pathlib import Path

root = Path('sandbox/project')

files = ['README.md', 'main.py', 'data/users.csv']

for file in files:
    path = root / file
    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    path.touch(exist_ok=True)
````

## Exercice 3

Crée une archive ZIP :

project_backup.zip

à partir de :

project/

````python
from pathlib import Path
import shutil

path = Path('sandbox')

shutil.make_archive(
    path/'project_backup', 
    'zip',
    path/'project'
)
````

## Exercice 5

Crée un dossier :

sandbox/restored_project/

Puis extrait l'archive ZIP dans ce dossier.

````python
from pathlib import Path
import shutil

root = Path('sandbox')

(root/'restored_project').mkdir(
    parents=True,
    exist_ok=True
)

# Extraction

shutil.unpack_archive(
    root/'project_backup.zip',
    root/'restored_project'
)
````

## Exercice 7

Supprime :

project/

puis vérifie que les données sont toujours récupérables grâce à l'archive.

(Indice conceptuel : l'archive devient alors la seule copie restante.)

````python
from pathlib import Path

root = Path('sandbox')

shutil.rmtree((root/'project'), ignore_errors=True)

if (
    not (root/'project').exists() 
    and 
    (root/'project_backup.zip').exists()
):
    print("L'archive est la seule copie restante")
````
