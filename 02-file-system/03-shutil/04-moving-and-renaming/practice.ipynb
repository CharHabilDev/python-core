## Exercice 1

Crée la structure suivante :

```txt
sandbox/
└── files/
    ├── notes.txt
    └── report.txt
```

````python
from pathlib import Path

root = Path('sandbox/files')

root.mkdir(
    parents=True,
    exist_ok=True
)

(root/'notes.txt').touch(exist_ok=True)
(root/'report.txt').touch(exist_ok=True)
````

## Exercice 2

Affiche le contenu de :

```txt
files/
```

````python
from pathlib import Path

path = Path('sandbox/files')

for element in path.iterdir():
    relative_path = element.relative_to(path)
    print(relative_path)
````

## Exercice 3

Crée un dossier :

archives/

puis déplace :

notes.txt

vers :

archives/


````python
from pathlib import Path
import shutil

root = Path('sandbox')

(root/'archives').mkdir(
    parents=True,
    exist_ok=True
)

source = root / 'files' / 'notes.txt'
destination = root / 'archives'

shutil.move(source, destination)
````

## Exercice 4

Vérifie que :

- notes.txt n'existe plus dans files/
- notes.txt existe dans archives/


````python
from pathlib import Path

root = Path('sandbox')

print(f"notes.txt existe dans files/ ? {(root / 'files' / 'notes.txt').exists()}")
print(f"notes.txt existe dans archives/ ? {(root / 'archives' / 'notes.txt').exists()}")
````

## Exercice 5

Renomme :

report.txt

en :

annual_report.txt

avec `shutil.move()`.


````python
from pathlib import Path
import shutil

root = Path('sandbox')

shutil.move(
    root/'files'/'report.txt', 
    root/'files'/'annual_report.txt'
)
````

## Exercice 6

Affiche le contenu final de :

files/

et de :

archives/

````python
from pathlib import Path

root = Path('sandbox')

dossiers = ['files', 'archives']

for dossier in dossiers:
    print(f"Dossier {dossier}/")
    for element in (root/dossier).iterdir():
        print(element.name)

````

## Exercice 7

Crée :

sandbox/project/

avec quelques fichiers.



````python
from pathlib import Path

path = Path('sandbox/project')

path.mkdir(
    parents=True,
    exist_ok=True
)

files_to_create = ["main.py", "config.json", "README.md"]

for file in files_to_create:
    (path/file).touch(exist_ok=True)
````

Puis déplace tout le dossier vers :

sandbox/project_backup/

````python
from pathlib import Path
import shutil

source = Path('sandbox/project')
destination = Path('sandbox/project_backup')

shutil.move(source, destination)
````
