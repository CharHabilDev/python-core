## Exercice 6

Un script de sauvegarde doit :

1. récupérer un dossier de sauvegarde depuis une variable d'environnement ;
    - pour récupérer la variable d'environnement
        - le module `os` avec la fonction `os.getenv()` ou `os.environ`
    - pour manipuler et valider le chemin de dossier de stockage
        - le module `pathlib` avec `Path()`
2. construire le chemin du projet ;
    - le module `pathlib` avec `Path()`
3. copier le projet ;
    - pour definir le chemin du dossier et sa copie
        - le module `pathlib`
    - pour faire la copie
        - le module `shutil` avec la fonction `shutil.copytree()`
4. créer une archive ZIP.
    - definir les chemins
        - le module `pathlib`
    - création de l'archive
        - le module `shutil` avec la fonction `shutil.make_archive()`

Quel module utiliserais-tu pour chaque étape ?
