## Exercice 1

Crée la structure suivante :
```txt
sandbox/
└── project/
    ├── README.md
    ├── main.py
    └── data/
        └── users.csv
```

````python
from pathlib import Path

root = Path('sandbox/project')

files = ['README.md', 'main.py', 'data/users.csv']

for file in files:
    path = root / file
    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    path.touch(exist_ok=True)
````

## Exercice 7

Crée :

sandbox/temp/

avec quelques fichiers.

Ajoute une vérification :

- le dossier existe ;
- c'est bien un dossier.

Puis supprime-le avec :

```python
shutil.rmtree(...)
```

````python
from pathlib import Path

root = Path('sandbox/temp')

temp_files = ["cache.tmp", "session.lock", "download.part"]

for file in temp_files:
    path = root / file
    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    path.touch(exist_ok=True)


#### Vérification :

if root.exists() and root.is_dir():
    shutil.rmtree(root)
````
