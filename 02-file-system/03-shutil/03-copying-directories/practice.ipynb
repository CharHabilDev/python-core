## Exercice 1

Crée la structure suivante :

```text
sandbox/
└── project/
    ├── README.md
    ├── notes.txt
    └── data/
        └── users.csv
````

`````python
from pathlib import Path

root = Path("sandbox/project")

files = ['README.md', 'notes.txt', 'data/users.csv']

for file in files:
    path = root / file
    path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    (root/file).touch(exist_ok=True)
`````

## Exercice 3

Copie :

```text
project/
```

vers :

```text
project_backup/
```

`````python
from pathlib import Path
import shutil

root = Path("sandbox")

source = root / "project"
copy = root / "project_backup"

shutil.copytree(source, copy)
`````

## Exercice 7

Recommence avec :

```python
dirs_exist_ok=True
```

Observe le résultat.


`````python
from pathlib import Path
import shutil

root = Path("sandbox")

source = root / "project"
destination = root / "project_backup"

shutil.copytree(
    source, 
    destination,
    dirs_exist_ok=True
)
`````

## Exercice 8

1. À quoi sert `copytree()` ?
    - copier un dossier complet ainsi que tous ses sous-dossiers et fichiers
2. Quelle différence existe entre `copy()` et `copytree()` ?
    - `copy()` copie les fichiers
    - `copytree()` copie des dossiers entiers
3. Pourquoi la destination ne doit-elle pas exister par défaut ?
    - pour éviter l'écrasement accidentel de données existantes
4. À quoi sert `dirs_exist_ok=True` ?
    - autoriser la copie même si le dossier destination existe déjà
5. Dans quel cas utiliserais-tu `copytree()` dans un projet réel ?
    - copier un dossier contenant une base de données
    - pour tester le projet sans risque de produire des erreur
