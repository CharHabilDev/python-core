## Exercice 2

Dans le bloc `with` :

- écris :

```text
Bonjour Context Manager
```

- lis le contenu
- affiche le résultat


````python
import tempfile

with tempfile.TemporaryFile(mode="w+") as temp:
    temp.write('Bonjour Context Manager')

    temp.seek(0)
    print(temp.read())
````

## Exercice 5

Crée un :

```python
with tempfile.TemporaryDirectory() as temp_dir:
```

Transforme le chemin en :

```python
Path(...)
```

et affiche-le.


````python
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as temp_dir:
    path = Path(temp_dir) 
    print(path)
````

## Exercice 6

Dans ce dossier temporaire :

- crée `notes.txt`
- crée `data/`
- affiche le contenu du dossier


````python
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as temp_dir:
    path = Path(temp_dir) 

    (path/'data').mkdir(
        parents=True,
        exist_ok=True
    )

    (path/'notes.txt').touch(exist_ok=True)


    for element in path.iterdir():
        print(element)
````
