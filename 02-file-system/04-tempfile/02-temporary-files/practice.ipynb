## Exercice 4

Replace le curseur au début du fichier.

Lis le contenu.

Affiche le résultat.

````python
import tempfile

tmp = tempfile.TemporaryFile(mode='w+', encoding='utf-8')

tmp.write('Bonjour tempfile')

tmp.seek(0)
print(tmp.read())
````

## Exercice 5

Ferme le fichier temporaire.

Vérifie que l'opération réussit.

````python
import tempfile

tmp = tempfile.TemporaryFile(mode='w+', encoding='utf-8')

tmp.write('Bonjour tempfile')

tmp.close()

print(f"Fichier temporaire fermé ? {tmp.closed}")
````

Complète :

| Outil                | Particularité                          |
| -------------------- | ---------------------------------------|
| TemporaryFile()      | créer un fichier temporaire dont le nom n'est généralement pas destiné à être utilisé directement              |

## Exercice 8

Réponds avec tes propres mots :

1. À quoi sert `TemporaryFile()` ?
    - à créer un fichier temporaire anonyme
2. À quoi sert `NamedTemporaryFile()` ?
    - même chose que `TemporaryFile()` mais il possède un nom de fichier
3. Pourquoi utiliser `seek(0)` ?
    - pour replacer le curseur au début du fichier
4. Dans quel cas aurais-tu besoin d'un nom de fichier visible ?
    - pour faire un traitement avec le fichier
    - une bibliothèque exige un chemin de fichier réel
5. Quelle différence principale existe entre les deux objets ?
    - `TemporaryFile()` : fichier temporaire généralement sans nom exploitable
    - `NamedTemporaryFile()`: fichier temporaire possédant un chemin visible accessible via .name
