## Exercice 4

Dans ce dossier temporaire :

* crée un fichier `notes.txt`
* vérifie qu'il existe

````python
import tempfile
from pathlib import Path

temp_dir = tempfile.TemporaryDirectory()

path = Path(temp_dir.name)

(path/'notes.txt').touch(exist_ok=True)

print(f"`notes.txt` existe ? {(path/'notes.txt').exists()}")
````

## Exercice 5

Crée un sous-dossier :

```text
data/
```

dans le dossier temporaire.

Vérifie son existence.

````python
import tempfile
from pathlib import Path

temp_dir = tempfile.TemporaryDirectory()

path = Path(temp_dir.name)

(path/'notes.txt').touch(exist_ok=True)

(path/'data').mkdir(
    parents=True,
    exist_ok=True
)

print(f"`data` existe ? {(path/'data').exists()}")
````

## Exercice 6

Affiche tout le contenu du dossier temporaire avec :

```python
iterdir()

````python
import tempfile
from pathlib import Path

temp_dir = tempfile.TemporaryDirectory()

path = Path(temp_dir.name)

(path/'notes.txt').touch(exist_ok=True)

(path/'data').mkdir(
    parents=True,
    exist_ok=True
)

for element in path.iterdir():
    print(element)
````
