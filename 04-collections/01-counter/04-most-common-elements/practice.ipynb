## Exercice 1

Crée un Counter à partir de :

```python id="9rdu1r"
"banana"
```

Affiche :

```python id="um2q2d"
counter.most_common()
```


````python
from collections import Counter

counter = Counter("banana")
print(counter.most_common())
````

## Exercice 2

Crée un Counter à partir de :

```python id="7kr5eg"
["python", "java", "python", "c", "python", "java"]
```

Affiche :

```python id="wm8ks7"
counter.most_common(2)
```


````python
from collections import Counter

counter = Counter(["python", "java", "python", "c", "python", "java"])

print(counter.most_common(2))
````
