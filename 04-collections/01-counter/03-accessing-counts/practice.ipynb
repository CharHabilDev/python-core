## Exercice 1

Crée un Counter à partir de :

```python id="17n6kp"
["red", "blue", "red", "green"]
```

Puis affiche le nombre d'occurrences de `"red"`.


````python
from collections import Counter

counter = Counter(["red", "blue", "red", "green"])

print(counter['red'])
````
