# 04 - Rabbit Statistics

## Objective

Generate statistics for a rabbit farm.

## Instructions

You are given a list containing rabbit breeds.

Count how many rabbits belong to each breed.

Then:

- Display all breed counts.
- Display the most common breed.
- Display the top three breeds.

### Expected Concepts

- `Counter`
- Accessing counts
- `most_common()`


```python
from collections import Counter

rabbit_breeds = [
    "Rex", "Angora", "Rex", "Lionhead", "Angora", 
    "Rex", "Flemish Giant", "Lionhead", "Rex", "Angora", 
    "Flemish Giant", "Lionhead", "Rex", "Rex", "Angora"
]

counter = Counter(rabbit_breeds)

print("All breed counts: ")
print(counter)
print("\nThe most common breed: ", end='')
breed, _ = counter.most_common(1)[0]
print(breed)
print("\nThe top three breeds: ")
print(counter.most_common(3))
```
