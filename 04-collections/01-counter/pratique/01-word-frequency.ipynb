# 01 - Word Frequency

## Objective

Create a simple word frequency analyzer.

## Instructions

You are given a text represented as a list of words.

Count how many times each word appears.

Then:

- Display all frequencies.
- Display the most common word.
- Display the three most common words.

### Expected Concepts

- `Counter`
- `most_common()`
- `most_common(n)`

```python
from collections import Counter

words = [
    "To", "understand", "recursion", "you", "must", "first", 
    "understand", "recursion", "because", "clean", "code", 
    "is", "not", "just", "code", "that", "works", "but", 
    "code", "that", "can", "be", "read", "and", "understood"
]

counter = Counter(words)

print("All frequencies: ")
print(counter)

print("\nThe most common word: ", end='')
word, _= counter.most_common(1)[0]

print(word)

print("\nThe three most common words: ")
print(counter.most_common(3))
```
