# 02 - Vote Counter

## Objective

Analyze the results of a small election.

## Instructions

You are given a list containing votes for different candidates.

Count the number of votes received by each candidate.

Then:

- Display all vote counts.
- Determine the winner.
- Display the ranking of candidates from most to least votes.

### Expected Concepts

- `Counter`
- `most_common()`
- Accessing counts

```python
from collections import Counter

votes = [
    "Alice", "Bob", "Alice", "Charlie", "Bob", 
    "Alice", "Alice", "Charlie", "Dave", "Bob", 
    "Alice", "Dave", "Charlie", "Bob", "Alice"
]

counter = Counter(votes)

print('All vote counts : ')
print(counter)

print("\nThe winner: ", end='')
name, _ = counter.most_common(1)[0]
print(name)

print("\nThe ranking of candidates from most to least votes: ")
for candidate in counter.most_common():
    name, count = candidate
    print(f"{f'{name}':<8}: {count} votes")
```
