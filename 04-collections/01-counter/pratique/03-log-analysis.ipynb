# 03 - Log Analysis

## Objective

Analyze application logs.

## Instructions

You are given a list containing log levels:

```text
INFO
WARNING
ERROR
CRITICAL
```

Count how many times each level appears.

Then:

- Display the frequency of each level.
- Display the most common log level.
- Determine how many errors occurred.

### Expected Concepts

- `Counter`
- Accessing counts
- `most_common()`


````python
from collections import Counter

log_levels = [
    "INFO", "WARNING", "INFO", "ERROR", "INFO", 
    "WARNING", "CRITICAL", "ERROR", "INFO", "WARNING", 
    "INFO", "ERROR", "CRITICAL", "INFO", "WARNING"
]

counter = Counter(log_levels)

print("The frequency of each level: ")
print(counter)
print("\nThe most common log level: ", end='')
most_common, _ = counter.most_common(1)[0]
print(most_common)
print(f"\nHow many errors occurred: {counter['ERROR']}")
````
