# 04 - Rabbit Breeds Registry

## Objective

Store rabbit breeds without duplicates.

## Instructions

You are given rabbit breed records.

Each breed may appear multiple times.

Group rabbits by breed while keeping only unique rabbit names.

Then:

- Display all breeds.
- Display the rabbits of each breed.
- Display the number of rabbits per breed.

### Expected Concepts

- `defaultdict(set)`
- Unique values
- Grouping data

```python
from collections import defaultdict

rabbit_records = [
    ("Rex", "Panpan"),
    ("Angora", "Fluffy"),
    ("Rex", "Panpan"),
    ("Lionhead", "Boulette"),
    ("Rex", "Jeannot"),
    ("Angora", "Fluffy"),
    ("Flemish Giant", "Goliath")
]

print("All breeds: ", end='')
breeds = {breed for breed, _ in rabbit_records}
print(breeds)

print(f"\nThe rabbits of each breed: ")
rabbits_by_breed = defaultdict(set)
for breed, rabbit in rabbit_records:
    rabbits_by_breed[breed].add(rabbit)
print(dict(rabbits_by_breed))

print(f"\nThe number of rabbits per breed: ")
number_of_rabbit = {}
for key, value in rabbits_by_breed.items():
    number_of_rabbit[key] = len(value)
print(dict(number_of_rabbit))
```
