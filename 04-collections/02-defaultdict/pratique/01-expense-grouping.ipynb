# 01 - Expense Grouping

## Objective

Group expenses by category.

## Instructions

You are given a list of expenses.

Group all amounts by category.

Then:

- Display all categories.
- Display the expenses of each category.
- Display the total amount for each category.

### Expected Concepts

- `defaultdict(list)`
- Grouping data
- Iteration

```python
from collections import defaultdict

expenses = [
    ("Food", 12.50), ("Transport", 4.50), ("Food", 8.20),
    ("Entertainment", 15.00), ("Transport", 10.00), ("Food", 22.10),
    ("Utilities", 85.40), ("Entertainment", 7.50), ("Food", 5.00)
]

print(f"All categories: ", end='')
categories = {category for category, _ in expenses}
print(categories)

print(f"\nThe expenses of each category: ")

data = defaultdict(list)
for expense in expenses:
    category, amount = expense
    data[category].append(amount)

print(data)

print(f"\nTotal amount for each category: ")
new_data = {}
for key, value in data.items():
    new_data[key] = sum(value)

print(new_data)
```
