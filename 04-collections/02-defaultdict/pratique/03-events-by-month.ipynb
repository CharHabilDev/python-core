# 03 - Events by Month

## Objective

Organize events by month.

## Instructions

You are given a list of events and their month.

Group events by month.

Then:

- Display all months.
- Display the events for each month.
- Display the month containing the most events.

### Expected Concepts

- `defaultdict(list)`
- Grouping
- Data organization

```python
from collections import defaultdict, Counter

months = [
    "Janvier", 
    "Juillet", 
    "Octobre", 
    "Janvier", 
    "Juillet", 
    "Décembre"
]

events_names = [
    "Nouvel An", 
    "Fête Nationale", 
    "Halloween", 
    "Soldes d'hiver", 
    "Festival de musique", 
    "Noël"
]

events = list(zip(months, events_names))

print("All months: ",end='')
months_= {month for month, _ in events}
print(months_)

print(f"\nThe events for each month: ")
calendar = defaultdict(list)
for month, event in events:
    calendar[month].append(event)
print(dict(calendar))

print(f"\nThe month containing the most events: ", end='')
counter = Counter(months)
top_month, _ = counter.most_common(1)[0]
print(top_month)
```
