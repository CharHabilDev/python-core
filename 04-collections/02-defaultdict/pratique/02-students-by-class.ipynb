# 02 - Students by Class

## Objective

Organize students by classroom.

## Instructions

You are given a list containing classroom names and student names.

Group students by classroom.

Then:

- Display each classroom.
- Display all students in that classroom.
- Display the number of students per classroom.

### Expected Concepts

- `defaultdict(list)`
- Automatic initialization
- Grouping records

```python
from collections import defaultdict

registrations = [
    ("Classroom A", "Léa"), ("Classroom B", "Hugo"), ("Classroom A", "Théo"),
    ("Classroom C", "Chloé"), ("Classroom B", "Emma"), ("Classroom A", "Léa"),
    ("Classroom C", "Lucas"), ("Classroom B", "Nathan")
]

print("Each classroom: ", end='')
classrooms = {classroom for classroom, _ in registrations}
print(list(classrooms))

print("\nAll students in that classroom: ")
data = defaultdict(list)
for classroom, student in registrations:
    data[classroom].append(student)
print(data)

print(f"\nThe number of students per classroom: ")
new_data = {}
for key, value in data.items():
    new_data[key] = len(value)
print(new_data)
```
