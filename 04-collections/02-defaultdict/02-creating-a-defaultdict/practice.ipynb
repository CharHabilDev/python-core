# Exercices

## Exercice 1

Crée un :

```python
defaultdict(int)
```

Puis affiche une clé inexistante.


````python
from collections import defaultdict

counter = defaultdict(int)
print(counter['one'])
````


---

## Exercice 2

Crée un :

```python
defaultdict(list)
```

Puis affiche une clé inexistante.


````python
from collections import defaultdict

data = defaultdict(list)

print(data['orange'])
````
