# 02 - Browser History (LIFO)

## Objective

Use a deque as a history stack.

## Instructions

Create a deque containing:

```text
Google
GitHub
Python Docs
MDN
```

Then:

1. Display the history.
2. Simulate pressing the Back button.
3. Display the removed page.
4. Display the remaining history.
5. Visit a new page.
6. Display the updated history.

### Concepts Practiced

- append()
- pop()
- LIFO

````python
from collections import deque

history = deque(["Google", "GitHub", "Python Docs", "MDN"])

print(f"Display the history: ")
print(history)

print(f"\nSimulate pressing the Back button: ")
previous_page = history.pop()

print(f"\nDisplay the removed page: {previous_page}")

print(f"\nDisplay the remaining history: ")
print(history)

print(f"\nVisit a new page: ")
history.append('CS50 Manual')
print("Visited: CS50 Manual")

print(f"\nDisplay the updated history: ")
print(history)
````
