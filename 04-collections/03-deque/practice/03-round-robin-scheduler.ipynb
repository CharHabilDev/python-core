# 03 - Round-Robin Scheduler

## Objective

Use rotation to cycle through users.

## Instructions

Create a deque containing:

```text
Alice
Bob
Charlie
David
```

Then:

1. Display the initial order.
2. Rotate once.
3. Display the new order.
4. Rotate again.
5. Display the result.
6. Identify whose turn it is after each rotation.

### Concepts Practiced

- rotate()
- Circular processing

````python
from collections import deque

users = deque(["Alice", "Bob", "Charlie", "David"])

print(f"Display the initial order: ")
print(users)

print(f"\nRotate once: ")
print(f"Rotated....")
users.rotate(1)

print(f"\nDisplay the new order: ")
print(users)

print(f"\nRotate again: ")
print(f"Rotated....")
users.rotate(1)

print(f"\nDisplay the result: ")
print(users)

print(f"\nIdentify whose turn it is after each rotation: ")
print("""
Initial turn: Alice
After first rotation: David
After second rotation: Charlie
""")
````
