# 01 - Task Queue (FIFO)

## Objective

Use a deque as a task queue.

## Instructions

Create a deque containing:

```text
Task A
Task B
Task C
Task D
```

Then:

1. Display the queue.
2. Process the first task.
3. Display the processed task.
4. Display the remaining queue.
5. Add a new task.
6. Display the final queue.

### Concepts Practiced

- deque()
- append()
- popleft()
- FIFO

````python
from collections import deque

tasks = deque(["Task A", "Task B", "Task C", "Task D"])

print("Display the queue: ")
print(tasks)

print("\nProcess the first task: ")
process_task = tasks.popleft()
print(f"Processing {process_task}...")

print(f"\nDisplay the processed task: {process_task}")

print("\nDisplay the remaining queue: ")
print(tasks)

print("\nAdd a new task: ")
tasks.append('Task E')
print(f"Task successfully added.")

print("\nDisplay the final queue: ")
print(tasks)
````
