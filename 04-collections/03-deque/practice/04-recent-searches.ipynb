# 04 - Recent Searches

## Objective

Store only the most recent searches.

## Instructions

Create a deque with:

```python
maxlen=5
```

Add the searches:

```text
python
django
flask
fastapi
postgresql
docker
git
```

Then:

1. Display the final deque.
2. Determine which searches were removed.
3. Explain why they were removed.

### Concepts Practiced

- maxlen
- Automatic removal
- Recent history

````python
from collections import deque
searches = deque(maxlen=5)
searches.append("python")
searches.append("django")
searches.append("flask")
searches.append("fastapi")
searches.append("postgresql")
searches.append("docker")
searches.append("git")

#searches = deque(["python", "django", "flask", "fastapi", "postgresql", "docker", "git"], maxlen=5)

print("The final deque: ")
print(searches)

print(f"\nDetermine which searches were removed: python and django")

print(f"\nExplain why they were removed: ")
print(
"They were removed because the deque can store only 5 elements. "
"When new elements are added beyond this limit, the oldest elements are automatically removed."
)
````
