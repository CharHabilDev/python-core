# 04 - Medicine Inventory

## Objective

Create a `Medicine` namedtuple with the following fields:

```text
name
price
stock
```

### Tasks

1. Create several medicine records.
2. Display all medicine information.
3. Convert a medicine record to a dictionary using `_asdict()`.
4. Create an updated copy using `_replace()`.
5. Verify that the original record remains unchanged.

````python
from collections import namedtuple

Medicine = namedtuple(
    'Medicine', 
    ['name', 'price', 'stock']
)

medicines = [
    Medicine(name="Amoxicillin", price=12.50, stock=150),
    Medicine(name="Ibuprofen", price=8.99, stock=200),
    Medicine(name="Paracetamol", price=4.50, stock=350),
    Medicine(name="Metformin", price=18.25, stock=90)
]

print("\nDisplay all medicine information:")
print(f"{'Medicine Name':<15} | {'Price ($)':<9} | {'Stock':<5}")
print("-" * 37)
for medicine in medicines:
    print(f"{medicine.name:<15} | ${medicine.price:<8.2f} | {medicine.stock:<5}")

print("\nConvert a medicine record to a dictionary:")
medicine = medicines[0]
print(f"Before: {medicine}")
print("Convertion.....")
print(f"After: {medicine._asdict()}")

print("\nCreate an updated copy using `_replace()`")
print(f"Update stock: {medicine._replace(stock=100)}")

print("\nVerify that the original record remains unchanged")
print(medicine)

````
