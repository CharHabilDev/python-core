# 03 - Rabbit Registry

## Objective

Create a `Rabbit` namedtuple with the following fields:

```text
id
name
breed
gender
```

### Tasks

1. Create several rabbit records.
2. Display all rabbit information.
3. Display only a rabbit's breed.
4. Display only a rabbit's gender.
5. Use `_fields` to display the available fields.

````python
from collections import namedtuple

Rabbit = namedtuple(
    "Rabbit",
    ['id', 'name', 'breed', 'gender']
)

rabbit_1 = Rabbit(id=1, name="Bugs", breed="Gray Hare", gender="Male")
rabbit_2 = Rabbit(id=2, name="Lola", breed="Holland Lop", gender="Female")
rabbit_3 = Rabbit(id=3, name="Thumper", breed="Angora", gender="Male")

rabbits = [rabbit_1, rabbit_2, rabbit_3]

print("--- Rabbit Records ---")
for rabbit in rabbits:
    print(f"ID: {rabbit.id} | Name: {rabbit.name} | Breed: {rabbit.breed} | Gender: {rabbit.gender}")

print("\nDisplay only a rabbit's breed: ")
for rabbit in rabbits:
    print(f"{rabbit.id}. {rabbit.breed}")

print("\nDisplay only a rabbit's gender: ")
for rabbit in rabbits:
    print(f"{rabbit.id}. {rabbit.gender}")

print("\nAvailable fields in Rabbit: ")
print(*Rabbit._fields, sep='\n')
````
