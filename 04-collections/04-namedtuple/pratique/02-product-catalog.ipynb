# 02 - Product Catalog

## Objective

Create a `Product` namedtuple with the following fields:

```text
name
price
stock
```

### Tasks

1. Create several products.
2. Display all products.
3. Display the price of a product.
4. Display the stock of a product.
5. Identify the most expensive product.

````python
from collections import namedtuple


print("Create several products: ")
Product = namedtuple(
    "Product",
    ['name', 'price', 'stock']
)

product_1 = Product(name="Laptop", price=999.99, stock=15)
product_2 = Product(name="Smartphone", price=599.99, stock=30)
product_3 = Product(name="Headphones", price=149.99, stock=50)
product_4 = Product(name="Wireless Mouse", price=29.99, stock=120)
product_5 = Product(name="4K Monitor", price=349.50, stock=8)

inventory = [product_1, product_2, product_3, product_4, product_5]

print("\nDisplay all products: ")
for i,  product in enumerate(inventory, start=1):
    print(f"{i}. {product}")

print("\nDisplay the price of a product: ")
for i, product in enumerate(inventory, start=1):
    print(f"{i}. {product.price} €")

print("\nDisplay the stock of a product: ")
for i, product in enumerate(inventory, start=1):
    print(f"{i}. {product.stock}")

print("\nThe most expensive product: ")
most = max(inventory, key=lambda x: x.price)
print(most)
````
