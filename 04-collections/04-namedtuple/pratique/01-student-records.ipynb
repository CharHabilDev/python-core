# 01 - Student Records

## Objective

Create a `Student` namedtuple with the following fields:

```text
name
classroom
grade
```

### Tasks

1. Create the `Student` type.
2. Create three student records.
3. Display all information for the first student.
4. Display only the student's name.
5. Display only the student's grade.

````python
from collections import namedtuple

print("Create the `Student` type")
Student = namedtuple(
    'Student',
    ["name", "classroom", "grade"]
)

print(f"\nCreate three student records: ")
student1 = Student(name="Alice", classroom="Terminale A", grade=15.5)
student2 = Student(name="Thomas", classroom="Terminale B", grade=12.0)
student3 = Student(name="Yasmine", classroom="Terminale A", grade=18.0)

print(f"\nDisplay all information for the first student: ")
print(
    f"Name: {student1.name} | "
    f"Classroom: {student1.classroom} | "
    f"Grade: {student1.grade}"
)

print(f"\nDisplay only the student's name: ")
students = [student1, student2, student3]
for i, student in enumerate(students, start=1):
    print(f"Student {i}: {student.name}")

print(f"\nDisplay only the student's grade: ")
for i, student in enumerate(students, start=1):
    print(f"Student {i}: {student.grade}")
````
